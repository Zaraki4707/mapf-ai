## Problem Formulation

### 1. State Representation

```python
state = {
    'robots': [
        {
            'id': 1,
            'position': (x, y),
            'goal': (gx, gy),
            'path': [(x1, y1), (x2, y2), ...],
            'path_index': 0,
            'at_goal': False
        },
        # ... more robots
    ],
    'time_step': 0,
    'collisions': 0,
    'deadlock': False,
    'grid': GridEnvironment
}
```

### 2. Actions

- **Plan Paths**: Use A* (independent, cooperative, or CBS) to find paths
- **Move Robot**: Robot moves one step along planned path
- **Execute Step**: All robots move simultaneously (one step each)

### 3. Goal Test

```python
def is_goal(state):
    """All robots reached goals without conflicts"""
    return all(robot['at_goal'] for robot in state['robots'])
```

### 4. Path Cost

- **Makespan**: Time until last robot finishes = max(len(path) for all robots)
- **Flowtime**: Sum of all path lengths = sum(len(path) for all robots)
- Each move costs: 1 time unit
- Collision penalty: +10 per conflict
- Deadlock penalty: +100
